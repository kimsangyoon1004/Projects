```python
import pandas as pd
import numpy as np
df = pd.read_csv("NBA.csv")
```

```python
def Get_ID(S):
    if S == "MIA":
        return 1
    if S == "BOS":
        return 2
    if S == "MIL":
        return 3
    if S == "PHI":
        return 4
    if S == "TOR":
        return 5
    if S == "CHI":
        return 6
    if S == "BRK":
        return 7
    if S == "ATL":
        return 8
    if S == "CLE":
        return 9
    if S == "CHO":
        return 10
    if S == "NYK":
        return 11
    if S == "WAS":
        return 12
    if S == "IND":
        return 13
    if S == "DET":
        return 14
    if S == "ORL":
        return 15
    if S == "PHO":
        return 16
    if S == "MEM":
        return 17
    if S == "GSW":
        return 18
    if S == "DAL":
        return 19
    if S == "UTA":
        return 20
    if S == "DEN":
        return 21
    if S == "MIN":
        return 22
    if S == "NOP":
        return 23
    if S == "LAC":
        return 24
    if S == "SAS":
        return 25
    if S == "LAL":
        return 26
    if S == "SAC":
        return 27
    if S == "POR":
        return 28
    if S == "OKC":
        return 29
    if S == "HOU":
        return 30
   
```

```python
Mat = np.zeros((30, 30))
Q = np.zeros((30, 30))
for i in range(0, 1228):
    t1 = Get_ID(df.loc[i, "Team_Abbrev"]) - 1
    t2 = Get_ID(df.loc[i, "Opponent_Abbrev"]) - 1
    p1 = df.loc[i, "Team_Score"]
    p2 = df.loc[i, "Opponent_Score"]
    if p1 > p2:
        Mat[t1][t2] += p1 - p2
    else:
        Mat[t2][t1] += p2 - p1
for i in range(0, 30):
    Sum = 0
    for j in range(0, 30):
        Sum += Mat[j][i]
    for j in range(0, 30):
        Mat[j][i] /= Sum
        Q[j][i] = Mat[j][i] * 0.8 + 0.2 / 30
```

```python
V = np.ones(30)
V /= 30
for i in range(1000):
    V = np.matmul(Q, V)
print (np.argsort(-V)+1)
```
